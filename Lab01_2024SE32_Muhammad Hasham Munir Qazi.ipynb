## Task 6
What is the main difference between a tuple and a list?

**Answer (write in your own words):**

A list is *mutable*, so its items can be changed, added or removed after it is created (e.g. `my_list[0] = 99`). A tuple is *immutable*: once created it cannot be changed, and trying to do so raises a `TypeError`. Lists use `[ ]` and tuples use `( )`.

The demo below shows this.

```python
my_list = [1, 2, 3]
my_list[0] = 99
print("Q6: list after change ->", my_list)

my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 99
except TypeError as e:
    print("Q6: tuple cannot be changed ->", e)
```

## Task 7
Function: grab the domain from `"user@domain.com"`.

```python
def domain_get(email):
    """Return the part of an email address after the @ sign."""
    return email.split("@")[1]

print("Q7:", domain_get("user@domain.com"))
```

## Task 8
Function: return `True` if the word "dog" is in a string (case-insensitive).

```python
def find_dog(text):
    """Return True if 'dog' appears in text, ignoring upper/lower case."""
    return "dog" in text.lower()

print("Q8:", find_dog("Is there a dog here?"))
print("Q8:", find_dog("Is there a DOG here?"))
print("Q8:", find_dog("No pets here"))
```

## Task 9
Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(text):
    """Return how many times 'dog' occurs in text, ignoring case."""
    return text.lower().count("dog")

print("Q9:", count_dog("This dog runs faster than the other dog"))
print("Q9:", count_dog("Dog, DOG and dog"))
```

## Task 10
`caught_speeding(speed, is_birthday)`:
- speed ≤ 60 → "No Ticket"
- 61 to 80 → "Small Ticket"
- ≥ 81 → "Big Ticket"
- On your birthday, add 5 to the allowed limits.

```python
def caught_speeding(speed, is_birthday):
    """Return the ticket type for a given speed."""
    if is_birthday:
        speed = speed - 5      # birthday gives 5 mph extra allowance

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10: caught_speeding(60, False) ->", caught_speeding(60, False))
print("Q10: caught_speeding(65, True)  ->", caught_speeding(65, True))
print("Q10: caught_speeding(65, False) ->", caught_speeding(65, False))
print("Q10: caught_speeding(80, True)  ->", caught_speeding(80, True))
print("Q10: caught_speeding(85, True)  ->", caught_speeding(85, True))
print("Q10: caught_speeding(86, True)  ->", caught_speeding(86, True))
print("Q10: caught_speeding(81, False) ->", caught_speeding(81, False))
```

## Bonus Challenges (optional, from the Submission slide)

### Bonus 1: `count_dog` that also counts `"dog,"` and `"dog!"`

```python
import string

def count_dog_punct(text):
    """Count the word 'dog' even when punctuation is attached, e.g. 'dog,' or 'dog!'."""
    words = text.lower().split()
    return sum(1 for w in words if w.strip(string.punctuation) == "dog")

print("Bonus1:", count_dog_punct("Dog, dog! A dog? Hotdog is not one."))
```

### Bonus 2: `is_palindrome(word)` using slicing

```python
def is_palindrome(word):
    """Return True if word reads the same forwards and backwards."""
    word = word.lower()
    return word == word[::-1]

print("Bonus2:", is_palindrome("Level"), is_palindrome("Python"))
```
